## **Задача 1. Регрессия для kNN (1 балл)**

Рассмотрим алгоритм KNN для решения задачи регрессии. На лекциях мы говорили, что для нового объекта $u$ прогноз делается по формуле:

\begin{equation}
    \hat{y}(u) = \arg \min_{c\in\mathcal{R}} \sum_{j=1}^{k} w_j (y_u^{(j)}-c)^2
\end{equation}

где $c$ - 'усредненное' значение целевой переменной по соседям; $y_u^{(j)}$ - значение целевой переменной $j$-го соседа объекта $u$; $w_j$ - некоторый вес соседа; $\hat{y}(u)$ - прогноз для объекта $u$. Покажите, что в явном виде прогноз можно записать так:


\begin{equation}
    \hat{y}(u) = \frac{1}{\sum_{j=1}^{k} w_j } \sum_{j=1}^{k} w_j y_u^{(j)}.
\end{equation}

**Решение:**

Обозначим $L(c) = \sum_{j=1}^{k} w_j (y_u^{(j)}-c)^2 $. Нам нужно найти такое с, при котором $L(c)$ минимальна.
Возьмем производную по с: $L^{'}(c) = -2\sum_{j=1}^{k} w_j (y_u^{(j)}-c) = 2\sum_{j=1}^{k} w_j (c - y_u^{(j)})$. В точке минимума: $L^{'}(c) = 0$.

Следовательно: \begin{equation}
2\sum_{j=1}^{k} w_j (c - y_u^{(j)}) = 0
\end{equation}

\begin{equation}
\sum_{j=1}^{k} w_j (c - y_u^{(j)}) = 0
\end{equation}

\begin{equation}
\sum_{j=1}^{k} w_j c - \sum_{j=1}^{k} w_j  y_u^{(j)} = 0
\end{equation}

\begin{equation}
c\sum_{j=1}^{k} w_j  - \sum_{j=1}^{k} w_j  y_u^{(j)} = 0
\end{equation}

\begin{equation}
c\sum_{j=1}^{k} w_j  = \sum_{j=1}^{k} w_j  y_u^{(j)}
\end{equation}

\begin{equation}
c = \frac{\sum_{j=1}^{k} w_j  y_u^{(j)}}{\sum_{j=1}^{k} w_j}
\end{equation}

Проверим, что найденное значение действительно является минимумом.
Первая производная: $L'(c)=2\sum_{j=1}^{k}w_j(c-y_u^{(j)})$.
Возьмем вторую производную по $c$: $L''(c)=2\sum_{j=1}^{k}w_j$.
Поскольку веса соседей неотрицательны, $w_j \geq 0$, и хотя бы один вес положителен, то $\sum_{j=1}^{k}w_j>0$. Следовательно, $L''(c)=2\sum_{j=1}^{k}w_j>0$.

Так как в найденной точке первая производная равна нулю, а вторая производная положительна, то найденная точка является точкой минимума.

То есть:
$$
\boxed{
\hat{y}(u) = \frac{1}{\sum_{j=1}^{k} w_j } \sum_{j=1}^{k} w_j y_u^{(j)}}.
$$

Ч.т.д.

## **Задача 3. Аналитическое решение (1 балл)**

Найдите веса линейной регрессии из предыдущей задачи с помощью аналитической формулы.

**Решение:**

Аналитическая формула для линейной регрессии: $w = (X^TX)^{-1}X^T y$.

Нам нужно найти два коэффициента:$$ w = \begin{pmatrix} w_0 \\ w_1 \end{pmatrix}$$

Первый коэффициент $w_0$ - свободный член. Чтобы он появился в матричной записи, к исходному $X$ добавим столбец единиц:

$$ X = \begin{pmatrix} 1 & 0 \\ 1& 0.5 \\ 1 & 1\\ 1 & 1.5 \end{pmatrix} $$

Теперь первый столбец соответствует коэффициенту $w_0$, второй столбец соответствует коэффициенту $w_1$.

Транспонируем матрицу:

$$X^T = \begin{pmatrix} 1 & 1 & 1 & 1 \\ 0 & 0.5 & 1 & 1.5 \end{pmatrix} $$

Тогда:

$$X^TX = \begin{pmatrix} 1 & 1 & 1 & 1 \\ 0 & 0.5 & 1 & 1.5 \end{pmatrix} \begin{pmatrix} 1 & 0 \\ 1& 0.5 \\ 1 & 1\\ 1 & 1.5 \end{pmatrix} = \begin{pmatrix} 4 & 3\\ 3 & 3.5 \end{pmatrix}$$

Найдем обратную матрицу:

$$(X^TX)^{-1} = \frac {1}{4 \cdot 3.5 - 3 \cdot 3} \begin{pmatrix} 3.5 & -3\\ -3 & 4 \end{pmatrix} = \frac {1}{5} \begin{pmatrix} 3.5 & -3\\ -3 & 4 \end{pmatrix} $$

Найдем $X^Ty$:

$$ X^T y = \begin{pmatrix} 1 & 1 & 1 & 1 \\ 0 & 0.5 & 1 & 1.5 \end{pmatrix} \begin{pmatrix} 0.6 \\ 1.4 \\ 2.7 \\ 3.3 \end{pmatrix} = \begin{pmatrix} 8 \\ 8.35 \end{pmatrix}$$

Подставляем все в формулу:

$$ w = (X^TX)^{-1}X^T y = \frac {1}{5} \begin{pmatrix} 3.5 & -3\\ -3 & 4 \end{pmatrix}\begin{pmatrix} 8 \\ 8.35 \end{pmatrix} = \begin{pmatrix} 0.59 \\ 1.88 \end{pmatrix} $$

Следовательно, $$\boxed{w_0 = 0.59, w_1 = 1.88}$$

$$\boxed{\hat{y} = 0.59 + 1.88x_1}$$



## **Задача 4. Ridge-регрессия (1 балл)**

Рассмотрим линейную регрессию с $L_2$ регуляризацией. Докажите, что аналитическое решение задается формулой:

\begin{equation}
    w = (X^TX+ \alpha I)^{-1}X^Ty,
\end{equation}

где $\alpha$ - коэффициент регуляризации; $I$ - единичная матрица.

**Решение:**

Линейная регрессия с $L_2$ регуляризацией:

$$L_{\alpha}(w) = L(w) + \alpha R_2 = (Xw - y)^T(Xw - y) + \alpha w^T w$$


$$L_{\alpha}(w)=(w^T X^T-y^T)(Xw-y)+\alpha w^T w =$$

$$=w^T X^T Xw-w^T X^T y-y^T Xw+y^T y+\alpha w^T w$$

Заметим, что $w^T X^T y$ и $y^T Xw$ — скаляры. Транспонирование скаляра не меняет его.

$$
(w^T X^T y)^T=((w^T X^T)y)^T=y^T(w^T X^T)^T=y^T Xw\Leftrightarrow w^T X^T y=y^T Xw$$

$$L_{\alpha}(w)=w^T X^T Xw-2y^T Xw +y^T y+\alpha w^T w$$

$$L_{\alpha}(w)=w^T X^T Xw-2y^T Xw +y^T y+\alpha w^T Iw$$

Нам необходимо минимизировать функцию ошибок по $w$, поэтому найдем градиент функции и приравняем его к $0$.

Правила дифференцирования матриц:

- $\frac{\partial}{\partial w}(w^T A w)=2Aw$
(если А симметрична)

- $\frac{\partial}{\partial w}(b^T w)=b$


- $\frac{\partial}{\partial w}(c)=0$

<br>

$(X^T X)^T=X^T X\Leftrightarrow$
матрица симметрична, значит, для первого слагаемого можно применить первое правило. Для второго слагаемого применяем второе правило для линейного выражения. Третье слагаемое вообще не зависит от $w$, поэтому его производная равна 0. Единичная матрица симметрична, следовательно, для четвертого слагаемого тоже применимо первое правило.

$$\nabla L_{\alpha}(w)=2X^T Xw-2X^T y+2\alpha w=0$$

$$X^T Xw-X^T y+\alpha w=0$$


$$X^T Xw+\alpha w= X^T y$$

$$(X^T X+\alpha I) w= X^T y$$

$$
\boxed{
w=(X^T X+\alpha I)^{-1}X^T y
}
$$

## **Задача 5. Геометрический смысл (1 балл)**

Докажите, что вектор предсказаний $Xw^*$ — ортогональная проекция $y$ на пространство столбцов $X$. Почему предсказания единственны, даже когда параметры не единственны? $w^*$ - это вектор оптимальных весов линейной регрессии.

**Решение**:

$$ L = (Xw - y) ^T (Xw - y) $$

$$\frac{\partial L} {\partial w} = 0$$

$$2X^T(Xw - y) = 0$$

Для оптимальных весов $w^*$:

$$2X^T(Xw^* - y) = 0$$

$$X^T(Xw^* - y) = 0$$

$$X^T(y - Xw^*) = 0$$

Вектор $y - Xw^*$ - это вектор ошибок модели, то есть разница между настоящими значениями $y$ и предсказаниями $Xw^*$. Равенство $X^T(y - Xw^*) = 0$ означает, что вектор $y - Xw^*$ ортогонален каждому столбцу матрицы $X$. Следовательно, он ортогонален всему пространству, натянутому на столбцы $X$. При этом сам вектор $Xw^*$ является линейной комбинацией столбцов $X$, значит, лежит в пространстве этих столбцов. Таким образом, $Xw^*$ - ортогональная проекция $y$ на пространство столбцов $X$.

Ортогональная проекция вектора на заданное пространство единственна. Так как мы уже доказали, что любое оптимальное предсказание $Xw^*$ является ортогональной проекцией $y$ на пространство столбцов $X$, два разных оптимальных набора весов не могут давать разные предсказания. Поэтому если существует два оптимальных решения $w^*_1$ и $w^*_2$, обязательно чтобы выполнялось: $Xw^*_1 = Xw^*_2$. Так, параметр $w^*$ может быть не единственным, но при этом вектор предсказаний $Xw^*$ всегда единственен.

## **Задача 6. Вероятностная интерпретация (1 балл)**

Пусть

$$
y_i=x_i^\top w+\varepsilon_i,\qquad
\varepsilon_i\overset{\mathrm{i.i.d.}}{\sim}\mathcal N(0,\sigma^2).
$$

При фиксированной $\sigma^2>0$ покажите, что оценка максимального правдоподобия для $w$ совпадает с решением метода наименьших квадратов.

**Решение:**

По условию, ошибка имеет нормальное распределение, значит, при фиксированном $x_i$ величина $y_i$ тоже имеет нормальное распределение:

$y_i | x_i \sim \mathcal{N}(x^T_iw, \sigma ^2)$ $(E(\varepsilon_i) = 0 \Rightarrow E(y_i | x_i) = x^T_iw$).

Для нормального распределения плотность имеет вид:

$$p(y_i | x_i, w) = \frac{1}{\sqrt{2\pi\sigma ^2}}\exp \left(-\frac{(y_i - x_i^Tw)^2}{2\sigma ^ 2} \right)$$

Т.к. ошибки независимы по условию:

$$L(w) = \prod_{i=1}^{n} p(y_i | x_i, w) = \prod_{i=1}^{n}\frac{1}{\sqrt{2\pi\sigma ^2}}\exp \left(-\frac{(y_i - x_i^Tw)^2}{2\sigma ^ 2} \right)$$
<br>

$$L(w) = (2\pi\sigma^2)^{-\frac{n}{2}}\exp\left(-\frac{1}{2\sigma^2}\sum_{i=1}^n(y_i - x_i^Tw)^2\right) \rightarrow \max_w$$

Максимизировать $L(w)$ эквивалентно максимизировать $\ln L(w)$:

$$\ln L(w) = -\frac{n}{2}\ln(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n(y_i - x_i^Tw)^2$$

Первое  слагаемое от $w$ не зависит, $\large -\frac{1}{2\sigma^2}$ является константой при фикс. $\sigma^2>0$,

следовательно, максимизация $\ln L(w)$ эквивалентна минимизации $ \sum_{i=1}^n(y_i - x_i^Tw)^2$
(чтобы максимизировать выражение со знаком минус, необходимо минимизировать саму сумму). А это и есть задача метода наименьших квадратов: подобрать параметры линейной модели так, чтобы минимизировать сумму квадратов ошибок. Следовательно, метод максимального правдоподобия и МНК решают одну и ту же задачу. **Что и требовалось доказать**.

## **Задача 8. Производная функции потерь (1 балл)**

Рассмотрим модель логистической регрессии с функцией потерь:

\begin{equation}
    L = - \frac{1}{n} \sum_{i=1}^{n} y_i \log \hat{y}_{i} + (1-y_i) \log (1 - \hat{y}_{i}),
\end{equation}

\begin{equation}
    \hat{y}_{i} = \sigma(z_i),
\end{equation}

\begin{equation}
    z_{i} = x_i^{T}w.
\end{equation}

где $n$ - количество объектов в выборке; $\hat{y}_{i}$ - прогноз модели; $w$ - веса модели; Покажите, что градиент вычисляется по формуле:

\begin{equation}
    \frac{\partial L}{\partial w} = - \frac{1}{n} \sum_{i=1}^{n} x_i (y_i - \hat{y}_{i}).
\end{equation}

Как это выражение будет выглядеть в матричной форме?

**Решение:**

Чтобы не таскать сумму, сначала возьмем функцию потерь только для объекта $i$:

$$L_i = -[ y_i \log \hat{y}_{i} + (1-y_i) \log (1 - \hat{y}_{i})]$$.

Тогда: $$L = \frac{1}{n} \sum_{i = 1}^{n} L_i$$
<br>
По правилу производной сложной функции:

$$\frac{\partial L_i}{\partial w} = \frac{\partial L_i}{\partial \hat{y_i}} \cdot \frac{\partial \hat{y_i}}{\partial z_i} \cdot \frac{\partial z_i}{\partial w}$$
<br>
Найдем каждую из трех производных:

$$ \frac{\partial L_i}{\partial \hat{y_i}} = -\frac{y_i}{\hat{y_i}} + \frac{1-y_i}{1-\hat{y_i}}$$
<br>
$$\frac{\partial \hat{y_i}}{\partial z_i} =  \hat{y_i}(1 - \hat{y_i})$$
<br>
$$\frac{\partial z_i}{\partial w} = x_i$$
<br>

Подставим:

$$\frac{\partial L_i}{\partial w} = \left(-\frac{y_i}{\hat{y_i}} + \frac{1-y_i}{1-\hat{y_i}}\right)\hat{y_i}(1 - \hat{y_i})x_i = -x_i(y_i - \hat{y_i})$$
<br>
$$L = \frac{1}{n} \sum_{i = 1}^{n} L_i ⇒ \boxed{\frac{\partial L}{\partial w} = -\frac{1}{n}\sum_{i=1}^{n}x_i(y_i - \hat{y_i})}$$

Запишем данное выражение в матричной форме.
Пусть вектор истинных ответов:

$$y = \begin{pmatrix} y_1 \\. \\. \\. \\y_n \end{pmatrix}$$

А вектор предсказаний:

$$\hat{y}= \begin{pmatrix} \hat{y_1} \\. \\. \\. \\\hat{y_n} \end{pmatrix}$$

## **Задача 9. От логарифма шансов к вероятности (1 балл)**

Из предположения

$$
\log\frac{p(x)}{1-p(x)}=x^\top w
$$

выведите формулу $p(x)$. Как интерпретировать увеличение одного признака на единицу при фиксированных остальных? Как эта формула связана с логистической регрессией?

**Решение:**
$$
\log\frac{p(x)}{1-p(x)}=x^\top w
$$
<br>
$$\frac{p(x)}{1-p(x)} = e^{x^Tw}$$
<br>
$$p(x) = e^{x^Tw}(1-p(x)) = e^{x^Tw} - e^{x^Tw}p(x)$$
<br>
$$p(x) + e^{x^Tw}p(x) = e^{x^Tw}$$
<br>
$$p(x)(1 + e^{x^Tw}) = e^{x^Tw}$$
<br>
$$p(x) = \frac{e^{x^Tw}}{1 + e^{x^Tw}} = \frac{1}{1 + e^{-x^Tw}} = \sigma(x^Tw)$$
<br>


Пусть мы увеличиваем признак $x_j$ на $1$, а все остальные признаки оставляем неизменными.

Раскроем:

$$
x^T w = w_1x_1 + \ldots + w_jx_j + \ldots + w_dx_d.
$$

Если

$$
x_j \rightarrow x_j + 1,
$$

то соответствующая часть станет:

$$
w_j(x_j+1)=w_jx_j+w_j.
$$

Значит, величина $x^T w$ увеличится на

$$
\boxed{w_j}.
$$

Так как $$\frac{p(x)}{1-p(x)} = e^{x^T w},$$
<br>
после увеличения $x_j$ на единицу получаем:

$$
e^{x^T w + w_j} = e^{x^T w}e^{w_j}.
$$

Следовательно, шансы умножаются на $e^{w_j}$.

Поэтому при $w_j>0$ вероятность принадлежности объекта $x$ к классу 1 увеличивается, при $w_j<0$ - уменьшается, а при $w_j=0$ не изменяется.



Из исходного предположения в условии мы в итоге получили:

$$
p(x)=\frac{1}{1+e^{-x^T w}}
=\sigma(x^T w).
$$

А это и есть формула вероятностного предсказания логистической регрессии.

## **Задача 10. Вывод функции потерь (1 балл)**

Пусть $y_i\in\{0,1\}$ и $P(y_i=1\mid x_i)=\sigma(x_i^\top w)$. Выведите отрицательное логарифмическое правдоподобие и покажите, что оно совпадает с бинарной кросс-энтропией. Найдите и исправьте все ошибки в решении ниже. Укажите где и в чем были ошибки.

**Решение с ошибками:** Предположим условную независимость наблюдений при фиксированных признаках и $w$. Обозначим $p_i=\sigma(x_i^Tw)$. Для бернуллиевской случайной величины
$$P(y_i\mid x_i,w)=p_i^{y_i},\qquad y_i\in\{0,1\}.$$
Тогда правдоподобие выборки и его логарифм равны
$$\mathcal L(w)=\prod_{i=1}^np_i^{y_i},$$
$$\log\mathcal L(w)=\sum_{i=1}^n\left[y_i\log p_i\right].$$
Отрицательное логарифмическое правдоподобие:
$$\boxed{-\log\mathcal L(w)=-\sum_{i=1}^n\left[y_i\log p_i\right].}$$
Это **суммарная бинарная кросс-энтропия**. После деления на $n$ получаем среднюю бинарную кросс-энтропию; положительный постоянный множитель не меняет множество минимумов.

**Исправленное решение:**

Нам дано: $$p_i = P(y_i=1\mid x_i)=\sigma(x_i^\top w)$$

Поскольку $y_i \in \{0, 1\}$, возможны два случая:

$$P(y_i|x_i, w) = \begin{cases}
p_i, & y_i=1,\\
1-p_i, & y_i=0.
\end{cases}
$$

Оба случая можно объединить в одну формулу:

$$\boxed{P(y_i|x_i, w) = p_i^{y_i}(1 - p_i)^{1-y_i}}$$
<br>
А в предложенном решении написано только: $P(y_i\mid x_i,w)=p_i^{y_i} \Rightarrow $ ошибка, пропущен множитель $(1 - p_i)^{1-y_i}$.

Дальше, поскольку наблюдения условно независимы, правдоподобие всей выборки будет иметь следующий вид:
$$\mathcal L(w)=\prod_{i=1}^np_i^{y_i}(1 - p_i)^{1-y_i}$$

Берем логарифм, произведение превращается в сумму:
$$\log\mathcal L(w)=\sum_{i=1}^n\left[y_i\log p_i + (1-y_i)\log(1-p_i)\right].$$

Значит, отрицательное логарифмическое правдоподобие будет иметь вид:

$$\boxed{-\log\mathcal L(w)=-\sum_{i=1}^n\left[y_i\log p_i + (1-y_i)\log(1-p_i)\right].}$$

Это суммарная бинарная кросс-энтропия. После деления на  n  получаем среднюю бинарную кросс-энтропию.

То есть ошибки в предложенном решении связаны между собой: в вероятности Бернулли пропущен множитель $(1-p_i)^{1-y_i}$, поэтому он затем отсутствует и в правдоподобии, и в логарифме правдоподобия, и в итоговой кросс-энтропии.
